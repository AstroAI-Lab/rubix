# Load SSP template via custom config

This shows how to use a custom configuration to load an SSP template that is stored under some file location on your disk.

```python
# NBVAL_SKIP
config_miles = {
        "name": "MILES",
        "format": "HDF5",
        "source": "https://github.com/geckos-survey/ngist/",
        "file_name": "MILES_template.h5",
        "fields": {
            "age": {
                "name": "age",
                "units": "Gyr",
                "in_log": False
            },
            "metallicity": {
                "name": "metallicity",
                "units": "",
                "in_log": False
            },
            "wavelength": {
                "name": "wavelength",
                "units": "Angstrom",
                "in_log": False
            },
            "flux": {
                "name": "flux",
                "units": "Lsun/Angstrom",
                "in_log": False
            }
        }
    }
```

```python
# NBVAL_SKIP
from rubix.spectra.ssp.grid import HDF5SSPGrid
ssp_miles = HDF5SSPGrid.from_file(config_miles, file_location="../rubix/spectra/ssp/templates")
ssp_miles
```

# Compare the h5 grid with the raw fits file as sanity check

```python
import os
import re
import numpy as np
from astropy.io import fits
import matplotlib.pyplot as plt
```

```python
test_fits_file = '/home/annalena/rubix/rubix/spectra/ssp/templates/MILES/Mch1.30Zp0.40T14.0000_iTp0.00_baseFe_linear_FWHM_2.51.fits'
```

```python
with fits.open(test_fits_file) as hdul:
    fits_flux = hdul[0].data
    header = hdul[0].header
    crval = header['CRVAL1']
    cdelt = header.get('CDELT1', header.get('CD1_1')) 
    fits_wave = crval + np.arange(len(fits_flux)) * cdelt
```

```python
pattern = re.compile(r'Z([pm]\d+\.\d+)T(\d+\.\d+)')
match = pattern.search(os.path.basename(test_fits_file))

Z_SUN = 0.012
z_str = match.group(1).replace('p', '+').replace('m', '-')
met_log = float(z_str)
target_Z = Z_SUN * (10 ** met_log)
target_age = float(match.group(2))

print(f"Target Age: {target_age} Gyr")
print(f"Target Z (Absolute): {target_Z}")
```

```python
age_idx = np.argmin(np.abs(np.array(ssp_miles.age) - target_age))
met_idx = np.argmin(np.abs(np.array(ssp_miles.metallicity) - target_Z))

print(f"Found matching HDF5 indices -> Age idx: {age_idx}, Met idx: {met_idx}")
```

```python
h5_flux = ssp_miles.flux[age_idx, met_idx, :] 
h5_wave = ssp_miles.wavelength
```

```python
h5_flux_np = np.array(h5_flux)

max_diff = np.max(np.abs(fits_flux - h5_flux_np))
print(f"Maximum absolute difference between FITS and loaded HDF5 flux: {max_diff:.4e}")
```

```python
plt.figure(figsize=(10, 5))

# Plot FITS data (thick solid line)
plt.plot(fits_wave, fits_flux, label='Original FITS', color='black', linewidth=4)

# Plot HDF5 data (dashed line on top)
plt.plot(h5_wave, h5_flux, label='Loaded HDF5 Grid', color='orange', linestyle='--', linewidth=2)

plt.title(f"Template Verification: Age = {target_age} Gyr, Z = {target_Z:.4f}")
plt.xlabel("Wavelength (Angstroms)")
plt.ylabel("Flux")
plt.legend()
plt.tight_layout()
plt.show()
```
